```python
from datasets import load_dataset
import os

ds = load_dataset("clane9/imagenet-100")

output_dir = "/imaging/nsakr/imagenet100_jpegs"

for split in ["train", "validation", "test"]:
    if split not in ds:
        print(f"⚠️ Split '{split}' not found, skipping.")
        continue
    
    for i, sample in enumerate(ds[split]):
        pil_image = sample["image"]
        label = sample["label"]

        # Make a folder per label index
        label_dir = os.path.join(output_dir, split, str(label))
        os.makedirs(label_dir, exist_ok=True)

        # Save as JPEG
        save_path = os.path.join(label_dir, f"{i}.jpg")
        pil_image.convert("RGB").save(save_path, format="JPEG")

    print(f"✅ Done saving {split} split.")
```
